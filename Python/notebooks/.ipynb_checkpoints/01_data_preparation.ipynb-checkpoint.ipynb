```python
def missing_percentage(df):
    result = pd.DataFrame({
        "missing_count": df.isnull().sum(),
        "missing_percentage": df.isnull().mean() * 100
    })
    
    return result.sort_values(
        "missing_percentage",
        ascending=False
    )
```

```python
display(missing_percentage(orders))
display(missing_percentage(order_items))
display(missing_percentage(products))
display(missing_percentage(refunds))
```

```python
order_summary = (
    order_items
    .groupby("order_id")
    .agg(
        total_items=("order_item_id", "count"),
        revenue=("price_usd", "sum"),
        cogs=("cogs_usd", "sum")
    )
    .reset_index()
)
```

```python
order_summary["profit"] = (
    order_summary["revenue"]
    - order_summary["cogs"]
)
```

```python
order_summary["profit_margin"] = np.where(
    order_summary["revenue"] != 0,
    (order_summary["profit"]
     / order_summary["revenue"]) * 100,
    0
)
```

```python
display(order_summary.head())
```

```python
orders_analysis = orders.merge(
    order_summary,
    on="order_id",
    how="left"
)
```

```python
print("Orders Analysis Shape:", orders_analysis.shape)

display(
    orders_analysis.head()
)
```

```python
orders_analysis["year"] = (
    orders_analysis["created_at"].dt.year
)

orders_analysis["month"] = (
    orders_analysis["created_at"].dt.month
)

orders_analysis["month_name"] = (
    orders_analysis["created_at"].dt.month_name()
)

orders_analysis["day_name"] = (
    orders_analysis["created_at"].dt.day_name()
)

orders_analysis["date"] = (
    orders_analysis["created_at"].dt.date
)
```

```python
display(
    orders_analysis[
        [
            "order_id",
            "user_id",
            "created_at",
            "total_items",
            "revenue",
            "cogs",
            "profit",
            "profit_margin"
        ]
    ].head()
)
```
